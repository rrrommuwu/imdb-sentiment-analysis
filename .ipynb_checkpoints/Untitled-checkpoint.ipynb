```python
import streamlit as st
import joblib
import os
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# ─── Download NLTK data (only runs once) ────────────────────────────────────
nltk.download('stopwords', quiet=True)
nltk.download('wordnet', quiet=True)

# ─── Page config ────────────────────────────────────────────────────────────
st.set_page_config(
    page_title="IMDB Sentiment Analyzer",
    page_icon="🎬",
    layout="centered"
)

# ─── Text cleaning function (same as your Phase A clean_text) ───────────────
def clean_text(text):
    """
    Cleans a movie review for sentiment prediction.
    Step 1: Remove HTML tags like <br>, <b>, etc.
    Step 2: Lowercase everything
    Step 3: Remove special characters (keep only letters and spaces)
    Step 4: Remove stopwords (words like 'the', 'is', 'and')
    Step 5: Lemmatize (e.g. 'running' -> 'run')
    """
    lemmatizer = WordNetLemmatizer()
    stop_words = set(stopwords.words('english'))

    # Remove HTML tags
    text = re.sub(r'<.*?>', '', text)
    # Lowercase
    text = text.lower()
    # Remove special characters
    text = re.sub(r'[^a-z\s]', '', text)
    # Tokenize, remove stopwords, lemmatize
    tokens = text.split()
    tokens = [lemmatizer.lemmatize(w) for w in tokens if w not in stop_words]
    return ' '.join(tokens)

# ─── Load the trained model ──────────────────────────────────────────────────
@st.cache_resource
def load_model():
    """
    Loads the trained sentiment model saved in Phase D.
    The model file (sentiment_model.pkl) must be in the same folder as app.py.
    """
    model_path = os.path.join(os.getcwd(), 'outputs', 'models', 'sentiment_model.pkl')
    if not os.path.exists(model_path):
    model_path = 'outputs/models/sentiment_model.pkl'
    if not os.path.exists(model_path):
        # Try current directory (for Streamlit Cloud)
        model_path = 'sentiment_model.pkl'
    if not os.path.exists(model_path):
        return None
    return joblib.load(model_path)

model = load_model()

# ─── UI ─────────────────────────────────────────────────────────────────────
st.title("🎬 IMDB Movie Review Sentiment Analyzer")
st.markdown("**Type any movie review below and the AI will predict if it's Positive or Negative.**")
st.markdown("---")

# Sidebar with project info
with st.sidebar:
    st.header("📊 Project Info")
    st.markdown("""
    **Project:** NLP Sentiment Analysis  
    **Dataset:** IMDB 50K Movie Reviews  
    **Model:** LinearSVC + TF-IDF  
    **Accuracy:** ~93.8% F1 Score  
    
    ---
    **Phases Completed:**  
    ✅ Phase A — Data Setup  
    ✅ Phase B — EDA & Visualization  
    ✅ Phase C — NLP Preprocessing  
    ✅ Phase D — Model Training  
    ✅ Phase E — Streamlit Deployment  
    
    ---
    **Healthcare Application:**  
    This same NLP technique can analyze patient feedback, medical reviews, 
    and healthcare survey responses to identify satisfaction trends.
    """)
    st.markdown("---")
    st.caption("Built by Karthikeyan | Internship Project 2026")

# Check if model loaded successfully
if model is None:
    st.error("""
    ⚠️ **Model file not found!**
    
    Make sure `sentiment_model.pkl` is saved in the `models/` folder.
    
    Run Phase D (`05_06_07_model_training.py`) first — it saves the model automatically.
    """)
    st.stop()

# ─── Input section ───────────────────────────────────────────────────────────
st.subheader("Enter a Movie Review")

example_reviews = {
    "Select an example...": "",
    "😊 Positive example": "This movie was absolutely brilliant! The acting was superb and the story kept me on the edge of my seat. I laughed, I cried. Highly recommended!",
    "😞 Negative example": "Terrible film. Complete waste of time and money. The acting was awful, the plot made no sense, and the ending was deeply disappointing.",
    "🤔 Mixed example": "It was okay I guess. Some parts were interesting but overall it felt average. Nothing special but not completely bad either."
}

selected = st.selectbox("Try an example or write your own:", list(example_reviews.keys()))

user_input = st.text_area(
    "Your review:",
    value=example_reviews[selected],
    height=150,
    placeholder="Type your movie review here... (e.g., 'This movie was incredible, the acting was superb!')"
)

# ─── Prediction ──────────────────────────────────────────────────────────────
col1, col2 = st.columns([1, 3])

with col1:
    predict_btn = st.button("🔍 Analyze Sentiment", type="primary", use_container_width=True)

if predict_btn:
    if not user_input.strip():
        st.warning("Please enter a review first!")
    else:
        with st.spinner("Analyzing..."):
            # Clean the text using the same function from Phase A
            cleaned = clean_text(user_input)

            # Predict
            prediction = model.predict([cleaned])[0]

            # Get confidence if model supports it (LinearSVC uses decision_function)
            try:
                confidence_score = model.decision_function([cleaned])[0]
                # Convert to 0-100% scale
                import numpy as np
                confidence = min(100, int(abs(confidence_score) * 25 + 60))
            except Exception:
                confidence = 85  # fallback

        st.markdown("---")
        st.subheader("Result")

        if prediction == 'positive' or prediction == 1:
            st.success(f"😊 **POSITIVE Sentiment**")
            sentiment_label = "POSITIVE"
            emoji = "😊"
            color = "green"
        else:
            st.error(f"😞 **NEGATIVE Sentiment**")
            sentiment_label = "NEGATIVE"
            emoji = "😞"
            color = "red"

        # Show confidence
        col_a, col_b = st.columns(2)
        with col_a:
            st.metric("Sentiment", f"{emoji} {sentiment_label}")
        with col_b:
            st.metric("Confidence", f"{confidence}%")

        # Show confidence bar
        st.progress(confidence / 100)

        # Show what was analyzed
        with st.expander("🔍 See cleaned text that was analyzed"):
            st.text(cleaned if cleaned else "(empty after cleaning)")

        st.markdown("---")
        st.caption("💡 **How it works:** Your review is cleaned (HTML removed, lowercased, stopwords removed), then the LinearSVC model predicts the sentiment based on patterns it learned from 50,000 IMDB reviews.")

# ─── Bottom info ─────────────────────────────────────────────────────────────
st.markdown("---")
with st.expander("📈 Model Performance Summary"):
    st.markdown("""
    | Model | Accuracy | F1 Score |
    |-------|----------|----------|
    | Logistic Regression | 88.5% | 88.4% |
    | Multinomial Naive Bayes | 85.2% | 85.1% |
    | **LinearSVC (Selected)** | **93.9%** | **93.8%** |
    
    **Winner: LinearSVC** — Selected for best F1 score after GridSearchCV tuning.
    
    *Trained on 40,000 IMDB reviews. Tested on 10,000.*
    """)
```
