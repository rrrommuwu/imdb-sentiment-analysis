```python
# ============================================================
# 01_data_loading.py — Phase A: Load & Clean IMDB Dataset
# Internship Project: NLP Sentiment Analysis (IMDB 50K)
# ============================================================

import pandas as pd
import re
import os

# ── Paths ────────────────────────────────────────────────────
DATA_PATH = "IMDB Dataset.csv"   # adjust if needed
OUT_PATH  = "data/imdb_cleaned.csv"

# ── clean_text ───────────────────────────────────────────────
def clean_text(text: str) -> str:
    """
    Remove HTML tags, non-alpha chars, lowercase, strip whitespace.
    """
    text = re.sub(r'<.*?>', ' ', text)          # strip HTML
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)   # keep letters only
    text = re.sub(r'\s+', ' ', text)            # collapse spaces
    return text.lower().strip()

def main():
    print("=" * 55)
    print("  Phase A — Data Loading & Cleaning")
    print("=" * 55)

    # 1. Load
    print(f"\n[1/4] Loading dataset from: {DATA_PATH}")
    df = pd.read_csv(DATA_PATH)
    print(f"      Shape          : {df.shape}")
    print(f"      Columns        : {list(df.columns)}")
    print(f"      Missing values : {df.isnull().sum().to_dict()}")

    # 2. Label encode
    print("\n[2/4] Encoding labels  (positive=1, negative=0)")
    df['label'] = (df['sentiment'] == 'positive').astype(int)
    print(f"      Class distribution:\n{df['label'].value_counts()}")

    # 3. Clean
    print("\n[3/4] Applying clean_text() …")
    df['clean_review'] = df['review'].apply(clean_text)
    df['review_len']   = df['clean_review'].apply(lambda x: len(x.split()))

    # 4. Save
    print(f"\n[4/4] Saving cleaned data -> {OUT_PATH}")
    os.makedirs("data", exist_ok=True)
    df.to_csv(OUT_PATH, index=False)

    print("\n✅ Done! Sample:")
    print(df[['clean_review', 'label', 'review_len']].head(3).to_string())
    print(f"\n   Saved {len(df):,} rows to {OUT_PATH}")

if __name__ == "__main__":
    main()
```
